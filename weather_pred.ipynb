```python
X=df.drop("Rain",axis=1)
y=df['Rain']
```

```python
from sklearn.model_selection import train_test_split
# train_test_split()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score ,classification_report ,confusion_matrix
lr=LogisticRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
y_pred
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train_sc=sc.fit_transform(X_train)
X_test_sc=sc.transform(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(random_state=10)
dt.fit(X_train_sc,y_train)

```

```python
y_pred_dt=dt.predict(X_test_sc)
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred_dt))
```

```python
from sklearn.svm import SVC
model_svm = SVC(kernel = 'linear')
```

```python
model_svm.fit(X_train_sc, y_train)
```

```python
y_pred
```

```python
y_pred_svc =model_svm.predict(X_test_sc)
accuracy_score(y_test,y_pred_svc)
```

```python
confusion_matrix(y_test,y_pred_svc)
```

```python
print(classification_report(y_test,y_pred_svc))
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc=RandomForestClassifier(random_state=10, n_estimators=100, max_depth=10,)
rfc.fit(X_train_sc,y_train)
```

```python
y_pred_rfc=rfc.predict(X_test_sc)
accuracy_score(y_test,y_pred_rfc)
```

```python
confusion_matrix(y_test,y_pred_rfc)
```

```python
print(classification_report(y_test,y_pred_rfc))
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier( n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    random_state=42,
    eval_metric='logloss')
xgb.fit(X_train_sc,y_train)
```

```python
y_pred_xgb = xgb.predict(X_test_sc)
y_pred_xgb
```

```python
accuracy_score(y_test,y_pred_xgb)
```

```python
confusion_matrix(y_test,y_pred_xgb)
```

```python
print(classification_report(y_test,y_pred_xgb) )
```

```python
from sklearn.metrics import accuracy_score

scores = {
    "Logistic Regression": accuracy_score(y_test, y_pred),
    "Decision Tree": accuracy_score(y_test, y_pred_dt),
    "SVM": accuracy_score(y_test, y_pred_svc),
    "Random Forest": accuracy_score(y_test, y_pred_rfc),
    "XGBoost": accuracy_score(y_test, y_pred_xgb)
}

print(scores)
```
